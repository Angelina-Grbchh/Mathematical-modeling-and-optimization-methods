```python
import numpy as np

c = np.array([2, 3, 0, -1, 0, 0])  
A = np.array([                   
    [2, -1, 0, -2, 1, 0],
    [3, 2, 1, -3, 0, 0],
    [-1, 3, 0, 4, 0, 1]
])
b = np.array([16, 18, 24])        

m, n = len(A), len(c)
T = np.zeros((m+1, n+m+1))

T[0, :n] = -c
for i in range(m):
    T[i+1, :n] = A[i]
    T[i+1, n+i] = 1
    T[i+1, -1] = b[i]

while min(T[0, :-1]) < -1e-6:
    j = np.argmin(T[0, :-1])
    ratios = [T[i,-1]/T[i,j] if T[i,j] > 1e-6 else np.inf for i in range(1, m+1)]
    i = np.argmin(ratios) + 1
    T[i] /= T[i,j]
    for k in range(m+1):
        if k != i: 
            T[k] -= T[k,j] * T[i]

print("Задача: z = 2x₁ + 3x₂ - x₄ → max")
print("Обмеження:")
print("2x₁ - x₂ - 2x₄ + x₅ = 16")
print("3x₁ + 2x₂ + x₃ - 3x₄ = 18") 
print("-x₁ + 3x₂ + 4x₄ + x₆ = 24")
print("x₁,x₂,x₃,x₄,x₅,x₆ ≥ 0\n")

print("Значення змінних в оптимальній точці:")
x_values = []
for j in range(n):
    value = 0
    for i in range(1, m+1):
        if abs(T[i,j] - 1) < 1e-6 and abs(T[0,j]) < 1e-6:
            value = T[i,-1]
    x_values.append(value)
    print(f"x_{j+1} = {value:.2f}")

print(f"\nОптимальне значення: z = {T[0,-1]:.2f}")

print("\nПеревірка цільової функції:")
x1, x2, x4 = x_values[0], x_values[1], x_values[3]
calc_z = 2*x1 + 3*x2 - x4
print(f"2({x1:.2f}) + 3({x2:.2f}) - ({x4:.2f}) = {calc_z:.2f}")
```
